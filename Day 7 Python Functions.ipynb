```python
# ============================
# SIMPLE BANKING SYSTEM
# ============================

def create_account(name, cnic):

    print("========== ACCOUNT CREATED ==========")
    print("Customer Name :", name)
    print("CNIC          :", cnic)
    print()


def deposit(balance, amount):

    balance = balance + amount

    return balance


def withdraw(balance, amount):

    if amount <= balance:

        balance = balance - amount

        return balance

    else:

        print("Insufficient Balance")

        return balance


def check_balance(balance):

    print("Current Balance : Rs.", balance)
    print()


def calculate_profit(balance):

    profit = balance * 0.10

    return profit


def print_receipt(name, balance):

    print("========== RECEIPT ==========")
    print("Customer :", name)
    print("Balance  : Rs.", balance)
    print("Thank You For Banking With Us")
    print("=============================")


# ============================
# Main Program
# ============================

customer_name = input("Enter Customer Name : ")

customer_cnic = input("Enter CNIC          : ")

create_account(customer_name, customer_cnic)

balance = 10000

check_balance(balance)

deposit_amount = float(input("Enter Deposit Amount : "))

balance = deposit(balance, deposit_amount)

check_balance(balance)

withdraw_amount = float(input("Enter Withdraw Amount : "))

balance = withdraw(balance, withdraw_amount)

check_balance(balance)

profit = calculate_profit(balance)

print("Annual Profit : Rs.", profit)

print()

print_receipt(customer_name, balance)
```

## Patient Registration

```python
def register_patient(name, age, disease):

    print("\nPatient Registered Successfully")
    print("Name :", name)
    print("Age :", age)
    print("Disease :", disease)


def appointment_fee():

    return 1500


def doctor_information():

    print("\nDoctor Name : Dr. Ahmad")
    print("Department : General Medicine")
    print("Timing : 9:00 AM to 5:00 PM")


def calculate_bill(fee, medicine):

    total = fee + medicine

    return total


def print_receipt(name, total):

    print("\nHospital Receipt")
    print("Patient :", name)
    print("Total Bill : Rs.", total)
    print("Thank You For Visiting Our Hospital")


patient_name = input("Enter Patient Name : ")

patient_age = int(input("Enter Age : "))

patient_disease = input("Enter Disease : ")

register_patient(patient_name, patient_age, patient_disease)

doctor_information()

medicine_bill = int(input("\nEnter Medicine Charges : "))

fee = appointment_fee()

total_bill = calculate_bill(fee, medicine_bill)

print_receipt(patient_name, total_bill)
```

```python
def create_account():

    name = input("Enter Account Holder Name : ")
    cnic = input("Enter CNIC : ")

    print("\nAccount Created Successfully")

    return name, cnic


def check_balance(balance):

    print("\nCurrent Balance : Rs.", balance)


def deposit(balance):

    amount = int(input("Enter Deposit Amount : "))

    balance = balance + amount

    print("Amount Deposited Successfully")

    return balance


def withdraw(balance):

    amount = int(input("Enter Withdraw Amount : "))

    if amount <= balance:

        balance = balance - amount

        print("Amount Withdraw Successfully")

    else:

        print("Insufficient Balance")

    return balance


def calculate_profit(balance):

    profit = balance * 0.10

    print("Annual Profit : Rs.", profit)


def receipt(name, balance):

    print("\nBank Receipt")

    print("Account Holder :", name)

    print("Available Balance : Rs.", balance)

    print("Thank You For Banking With Us")


name, cnic = create_account()

balance = 10000


while True:

    print("\nBank Management System")

    print("1. Check Balance")

    print("2. Deposit Money")

    print("3. Withdraw Money")

    print("4. Calculate Profit")

    print("5. Print Receipt")

    print("6. Exit")

    choice = int(input("Enter Your Choice : "))


    if choice == 1:

        check_balance(balance)


    elif choice == 2:

        balance = deposit(balance)


    elif choice == 3:

        balance = withdraw(balance)


    elif choice == 4:

        calculate_profit(balance)


    elif choice == 5:

        receipt(name, balance)


    elif choice == 6:

        print("\nThank You")

        break


    else:

        print("Invalid Choice")
```
