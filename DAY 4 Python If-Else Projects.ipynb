# Comparison Operators

Comparison operators are used to compare two values.

Python returns either:

- True
- False

## Operators Covered

* >
* <
* >=
* <=
* ==
*  !=

These operators are the foundation of decision making in Python.

```python
a = 30
b = 20


print(a>b)
print(a<b)
print(a<=b)
print(a>=b)
print(a==b)
```

# Understanding if-else Statement

The `if-else` statement allows Python to choose between two different outputs.

If the condition is True, the `if` block executes.

Otherwise, the `else` block executes.

```python
age = int(input("Enter age"))

if age >= 18:
 print("The Candidate is eligible")
else:
 print("not eligible")
```

# UNIVERSITY SCHOLARSHIP ELIGIBILITY CHECKER

```python
print("                    PYTHON ATM MACHINE")

bank_name = "Meezan Bank"
account_holder = input("Enter Account Holder Name : ")
account_number = input("Enter Account Number      : ")

balance = int(input("Enter Current Balance     : "))

print("1. Deposit")
print("2. Withdraw")

choice = int(input("Enter Your Choice (1 or 2): "))

if choice == 1:

    deposit = int(input("Enter Deposit Amount      : "))

    new_balance = balance + deposit

    print("Transaction Successful")
    print("--------------------------------------------------")
    print("Bank Name       :", bank_name)
    print("Account Holder  :", account_holder)
    print("Account Number  :", account_number)
    print("Previous Balance: Rs.", balance)
    print("Deposit Amount  : Rs.", deposit)
    print("Current Balance : Rs.", new_balance)

else:

    withdraw = int(input("Enter Withdraw Amount     : "))

    if withdraw <= balance:

        new_balance = balance - withdraw

        print("Transaction Successful")
        print("--------------------------------------------------")
        print("Bank Name       :", bank_name)
        print("Account Holder  :", account_holder)
        print("Account Number  :", account_number)
        print("Previous Balance: Rs.", balance)
        print("Withdraw Amount : Rs.", withdraw)
        print("Current Balance : Rs.", new_balance)

    else:

        print("Transaction Failed")
        print("Insufficient Balance")
        print("Available Balance : Rs.", balance)
        print("---------------------------------")
```

# ONLINE FOOD ORDERING SYSTEM

```python
print("\nONLINE FOOD ORDERING SYSTEM")
print("-------------------------------------------")

restaurant_name = "Cheezious"
city = "Lahore"

customer_name = input("Enter Customer Name           : ")
phone_number = input("Enter Phone Number            : ")
delivery_address = input("Enter Delivery Address        : ")

print("\nEnter Prices of Ordered Items")

pizza = int(input("Pizza (Rs.)                   : "))
burger = int(input("Burger (Rs.)                  : "))
fries = int(input("Fries (Rs.)                   : "))
zinger = int(input("Zinger Burger (Rs.)           : "))
sandwich = int(input("Sandwich (Rs.)                : "))
shawarma = int(input("Shawarma (Rs.)                : "))
drink = int(input("Cold Drink (Rs.)              : "))
ice_cream = int(input("Ice Cream (Rs.)               : "))
dessert = int(input("Dessert (Rs.)                 : "))
delivery_charges = int(input("Delivery Charges (Rs.)        : "))

subtotal = pizza + burger + fries + zinger + sandwich + shawarma + drink + ice_cream + dessert

discount = int(input("\nEnter Discount (%)            : "))
tax = int(input("Enter Sales Tax (%)           : "))

discount_amount = subtotal * discount / 100
amount_after_discount = subtotal - discount_amount

tax_amount = amount_after_discount * tax / 100

final_bill = amount_after_discount + tax_amount + delivery_charges

payment_method = input("\nPayment Method (Cash/Card)    : ")

print("\nFood Order Receipt")
print("-------------------------------------------")

print("Restaurant Name    :", restaurant_name)
print("City               :", city)

print()

print("Customer Name      :", customer_name)
print("Phone Number       :", phone_number)
print("Delivery Address   :", delivery_address)

print()

print("Pizza              : Rs.", pizza)
print("Burger             : Rs.", burger)
print("Fries              : Rs.", fries)
print("Zinger Burger      : Rs.", zinger)
print("Sandwich           : Rs.", sandwich)
print("Shawarma           : Rs.", shawarma)
print("Cold Drink         : Rs.", drink)
print("Ice Cream          : Rs.", ice_cream)
print("Dessert            : Rs.", dessert)

print()

print("Subtotal           : Rs.", subtotal)
print("Discount           :", discount, "%")
print("Discount Amount    : Rs.", discount_amount)
print("Amount After Discount : Rs.", amount_after_discount)
print("Sales Tax          :", tax, "%")
print("Tax Amount         : Rs.", tax_amount)
print("Delivery Charges   : Rs.", delivery_charges)

print()

print("Final Bill         : Rs.", final_bill)
print("Payment Method     :", payment_method)

if final_bill >= 3000:

    print("Special Offer      : Free Soft Drink")

else:

    print("Special Offer      : Thank you for your order")

print()
print("Order Status       : Confirmed")
print("Estimated Delivery : 30 Minutes")
print()
print("Thank you for ordering from")
print(restaurant_name)
```

# ELECTRICITY BILL CALCULATOR

```python
print("\nELECTRICITY BILL CALCULATOR")
print("-----------------------------------------------")

company_name = "LESCO - Lahore Electric Supply Company"

customer_name = input("Enter Customer Name            : ")
father_name = input("Enter Father's Name            : ")
phone_number = input("Enter Phone Number             : ")
city = input("Enter City                     : ")
address = input("Enter Address                  : ")
meter_number = input("Enter Meter Number             : ")
billing_month = input("Enter Billing Month            : ")

previous_reading = int(input("Enter Previous Meter Reading   : "))
current_reading = int(input("Enter Current Meter Reading    : "))
price_per_unit = int(input("Enter Price Per Unit (Rs.)     : "))

units_used = current_reading - previous_reading

electricity_charges = units_used * price_per_unit

fuel_adjustment = int(input("Enter Fuel Adjustment (Rs.)    : "))
meter_rent = int(input("Enter Meter Rent (Rs.)         : "))
service_charges = int(input("Enter Service Charges (Rs.)    : "))

subtotal = electricity_charges + fuel_adjustment + meter_rent + service_charges

tax = int(input("Enter GST (%)                  : "))

tax_amount = subtotal * tax / 100

final_bill = subtotal + tax_amount

print("\nElectricity Bill")
print("-----------------------------------------------")

print("Company Name        :", company_name)
print("Billing Month       :", billing_month)

print()

print("Customer Name       :", customer_name)
print("Father's Name       :", father_name)
print("Phone Number        :", phone_number)
print("City                :", city)
print("Address             :", address)
print("Meter Number        :", meter_number)

print()

print("Previous Reading    :", previous_reading)
print("Current Reading     :", current_reading)
print("Units Consumed      :", units_used)

print()

print("Price Per Unit      : Rs.", price_per_unit)
print("Electricity Charges : Rs.", electricity_charges)
print("Fuel Adjustment     : Rs.", fuel_adjustment)
print("Meter Rent          : Rs.", meter_rent)
print("Service Charges     : Rs.", service_charges)

print()

print("Subtotal            : Rs.", subtotal)
print("GST                 :", tax, "%")
print("GST Amount          : Rs.", tax_amount)

print()

print("Final Bill          : Rs.", final_bill)

if final_bill >= 10000:

    print("Bill Status         : High Electricity Bill")

else:

    print("Bill Status         : Normal Electricity Bill")

print()

print("Payment Due Date    : 15 August 2026")
print("Payment Status      : Pending")

print()

print("Thank you for using")
print(company_name)
```

# LOAN ELIGIBILITY CHECKER

```python
print("\nLOAN ELIGIBILITY CHECKER")
print("--------------------------------------------")

bank_name = "National Bank of Pakistan"
branch = "Main Branch, Lahore"

applicant_name = input("Enter Applicant Name           : ")
father_name = input("Enter Father's Name            : ")
cnic = input("Enter CNIC Number              : ")
phone = input("Enter Phone Number             : ")
city = input("Enter City                     : ")
occupation = input("Enter Occupation               : ")

age = int(input("Enter Age                      : "))
monthly_income = int(input("Enter Monthly Income (Rs.)     : "))
monthly_expense = int(input("Enter Monthly Expenses (Rs.)   : "))
loan_amount = int(input("Enter Required Loan (Rs.)      : "))
loan_years = int(input("Enter Loan Duration (Years)    : "))
credit_score = int(input("Enter Credit Score             : "))

net_income = monthly_income - monthly_expense

print("\nLoan Application Summary")
print("--------------------------------------------")

print("Bank Name          :", bank_name)
print("Branch             :", branch)

print()

print("Applicant Name     :", applicant_name)
print("Father's Name      :", father_name)
print("CNIC               :", cnic)
print("Phone Number       :", phone)
print("City               :", city)
print("Occupation         :", occupation)

print()

print("Age                :", age)
print("Monthly Income     : Rs.", monthly_income)
print("Monthly Expenses   : Rs.", monthly_expense)
print("Net Income         : Rs.", net_income)
print("Loan Amount        : Rs.", loan_amount)
print("Loan Duration      :", loan_years, "Years")
print("Credit Score       :", credit_score)

print()

if age >= 21 and monthly_income >= 50000 and credit_score >= 700:

    print("Loan Status        : APPROVED")

    if loan_amount <= 500000:
        interest_rate = 8

    elif loan_amount <= 1000000:
        interest_rate = 10

    else:
        interest_rate = 12

    interest = loan_amount * interest_rate / 100
    total_payment = loan_amount + interest
    monthly_installment = total_payment / (loan_years * 12)

    print("Interest Rate      :", interest_rate, "%")
    print("Interest Amount    : Rs.", interest)
    print("Total Payment      : Rs.", total_payment)
    print("Monthly Installment: Rs.", round(monthly_installment, 2))

else:

    print("Loan Status        : REJECTED")

    if age < 21:
        print("Reason             : Applicant must be at least 21 years old.")

    elif monthly_income < 50000:
        print("Reason             : Monthly income is below the required limit.")

    elif credit_score < 700:
        print("Reason             : Credit score is too low.")

print()
print("Application Status : Completed")
print("Thank you for using", bank_name)
```
