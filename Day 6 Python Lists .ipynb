```python
print("SHOPPING LIST MANAGEMENT SYSTEM")
print("-------------------------------------------")

shopping = []

total_items = int(input("How many items do you want to add? : "))

print()

for i in range(total_items):

    item = input("Enter Item Name : ")

    shopping.append(item)

print("\nCurrent Shopping List")
print("-------------------------------------------")

for item in shopping:
    print(item)

print("\nTotal Items :", len(shopping))

print("\nAdding a New Item...")

new_item = input("Enter New Item : ")

shopping.append(new_item)

print("\nShopping List")

for item in shopping:
    print(item)

print("\nInsert an Important Item")

important_item = input("Enter Important Item : ")

shopping.insert(0, important_item)

print()

for item in shopping:
    print(item)

print("\nRemove an Item")

remove_item = input("Enter Item Name to Remove : ")

if remove_item in shopping:

    shopping.remove(remove_item)

    print("Item Removed Successfully.")

else:

    print("Item Not Found.")

print("\nShopping List")

for item in shopping:

    print(item)

print("\nRemoving Last Item...")

shopping.pop()

for item in shopping:

    print(item)

print("\nSorting Items Alphabetically...")

shopping.sort()

for item in shopping:

    print(item)

print("\nReverse Order")

shopping.reverse()

for item in shopping:

    print(item)

print("\nFinal Shopping Summary")
print("-------------------------------------------")

print("Total Items :", len(shopping))

print("First Item :", shopping[0])

print("Last Item :", shopping[-1])

print("\nThank You For Using Shopping List Management System")
```

# SMART MINI CALCULATOR

```python
###### ==========================
#   SMART MINI CALCULATOR  
# ==========================

history = []


# --------------------------
# Calculator Functions
# --------------------------

def addition(num1, num2):
    return num1 + num2


def subtraction(num1, num2):
    return num1 - num2


def multiplication(num1, num2):
    return num1 * num2


def division(num1, num2):

    if num2 == 0:
        return "Division by Zero is Not Allowed"

    return num1 / num2


def modulus(num1, num2):
    return num1 % num2


def power(num1, num2):
    return num1 ** num2


def floor_division(num1, num2):
    return num1 // num2


# --------------------------
# Menu Function
# --------------------------

def menu():

    print("\n===================================")
    print("      SMART MINI CALCULATOR")
    print("===================================")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Modulus")
    print("6. Power")
    print("7. Floor Division")
    print("8. View History")
    print("9. Clear History")
    print("10. Exit")
    print("===================================")


# --------------------------
# Main Program
# --------------------------

while True:

    menu()

    choice = int(input("Enter Your Choice : "))

    if choice == 10:

        print("\nThank You For Using Smart Mini Calculator.")
        break

    elif choice == 8:

        print("\n========== HISTORY ==========")

        if len(history) == 0:
            print("No Calculation Found.")

        else:
            for item in history:
                print(item)

        continue

    elif choice == 9:

        history.clear()

        print("History Cleared Successfully.")

        continue

    elif choice >= 1 and choice <= 7:

        num1 = float(input("Enter First Number  : "))
        num2 = float(input("Enter Second Number : "))

        if choice == 1:

            answer = addition(num1, num2)

            operation = "+"

        elif choice == 2:

            answer = subtraction(num1, num2)

            operation = "-"

        elif choice == 3:

            answer = multiplication(num1, num2)

            operation = "*"

        elif choice == 4:

            answer = division(num1, num2)

            operation = "/"

        elif choice == 5:

            answer = modulus(num1, num2)

            operation = "%"

        elif choice == 6:

            answer = power(num1, num2)

            operation = "**"

        elif choice == 7:

            answer = floor_division(num1, num2)

            operation = "//"

        print("\nResult :", answer)

        history.append(f"{num1} {operation} {num2} = {answer}")

    else:

        print("Invalid Choice. Please Try Again.")
        
```
