### Dictionary with Loops

```python
student = {

    "Name": "Ali",
    "Age": 20,
    "City": "Lahore"

}

for key in student:

    print(key)
    
```

```python
student = {

    "Name": "Ali",
    "Age": 20,
    "City": "Lahore"

}

for key in student:

    print(student[key])
```

```python
student = {

    "Name": "Ali",
    "Age": 20,
    "City": "Lahore"

}

for key, value in student.items():

    print(key, ":", value)
```

```python
customer = {

    "Account Holder": "Ahmed",
    "Account Number": 123456,
    "Balance": 50000,
    "Account Type": "Saving"

}

print("Customer Information\n")

for key, value in customer.items():

    print(key, ":", value)
```

### summary of Data structure

#### Quick Summary

- **List** → Ordered, Changeable, Duplicates Allowed
- **Tuple** → Ordered, Not Changeable, Duplicates Allowed
- **Dictionary** → Ordered, Changeable, Stores Data as Key-Value Pairs, Duplicate Keys Not Allowed
- **Set** → Unordered, Changeable, Only Unique Values

### Mini project 

```python
def display_student(student):

    print("\nStudent Information")

    for key, value in student.items():

        print(key, ":", value)


student = {

    "Student ID": 101,
    "Name": "Ahmed Ali",
    "Department": "BS Mathematics",
    "Semester": "2nd",
    "CGPA": 3.82

}


subjects = (

    "Calculus",
    "Linear Algebra",
    "Programming Fundamentals",
    "English"

)


skills = {

    "Python",
    "MS Excel",
    "Python",
    "Communication",
    "Problem Solving",
    "Communication"

}


print("========== UNIVERSITY STUDENT MANAGEMENT SYSTEM ==========")

display_student(student)

print("\nRegistered Subjects")

for subject in subjects:

    print(subject)


print("\nStudent Skills")

for skill in skills:

    print(skill)


print("\nTotal Student Information :", len(student))

print("Total Subjects :", len(subjects))

print("Unique Skills :", len(skills))
```
