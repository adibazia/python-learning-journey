## What is Loop?

##### A loop allows us to execute the same block of code multiple times without writing it repeatedly.

Python mainly provides:

- for loop
- while loop

```python
for i in range(5):
    print("Adiba")
```

```python
for i in range(5):
    print("i")
```

```python
for i in range(5):
    print(i)
```

```python
for i in range(1,10):
    print(i)
```

## Mulitiplication Table using Loop (for)

```python
number = int(input("Enter a number:   "))

for i in range(1,11):
    print(number, "x" , i, "=", number * i)
```

```python
number = int(input("Enter a number:  "))
for i in range(1, 11):
    print(number, "x", i, "=", number * i)
```

```python
number = 1

while number <= 5:
    print(number)
    number = number + 1
```

## MUultiplication Table using Loop (while)

```python
print("MULTIPLICATION TABLE")
print("-------------------------")

number = int(input("Enter a Number: "))

i = 1

while i <= 10:
    print(number, "x", i, "=", number * i)
    i = i + 1

print("-------------------------")
print("Table Completed")
```

## BUS Ticket Booking System Loop (for)

```python
print("\nBUS TICKET BOOKING SYSTEM")
print("------------------------------------------")

company_name = "Daewoo Express"
route = "Lahore to Islamabad"

total_passengers = int(input("Enter Number of Passengers : "))

ticket_price = 2500
total_bill = 0

print()

for i in range(1, total_passengers + 1):

    print("Passenger", i)

    name = input("Enter Passenger Name : ")
    age = int(input("Enter Age            : "))
    gender = input("Enter Gender         : ")

    print()

    if age <= 10:

        fare = ticket_price * 0.50
        print("Child Discount Applied (50%)")

    elif age >= 60:

        fare = ticket_price * 0.70
        print("Senior Citizen Discount Applied (30%)")

    else:

        fare = ticket_price

    total_bill = total_bill + fare

    print("Ticket Price : Rs.", fare)
    print("------------------------------------------")

print()

print("BOOKING SUMMARY")
print("------------------------------------------")

print("Company Name      :", company_name)
print("Route             :", route)
print("Passengers        :", total_passengers)
print("Price Per Ticket  : Rs.", ticket_price)
print("Total Bill        : Rs.", total_bill)

print()

payment = input("Payment Method (Cash/Card) : ")

print()

if payment == "Cash":

    print("Payment Status : Cash Received")

else:

    print("Payment Status : Card Payment Successful")

print()

print("Booking Status : Confirmed")
print("Thank you for choosing", company_name)
```

## ATM PIN Verification loop (While)

```python
correct_pin = 7860

pin = int(input("Enter Your PIN : "))

while pin != correct_pin:

    print("Incorrect PIN. Please Try Again.")
    print()

    pin = int(input("Enter Your PIN : "))

print()
print("PIN Verified Successfully")
print("Welcome to Your Account")
```

# Nested Loop

```python
print("CINEMA HALL SEAT ARRANGEMENT")
print("--------------------------------")

for row in range(1, 4):

    print("Row", row)

    for seat in range(1, 5):

        print("Seat", seat)

    print()
```

```python
print("BUS SEAT ARRANGEMENT")
print("----------------------------")

row = 1

while row <= 3:

    print("Row", row)

    seat = 1

    while seat <= 4:

        print("Seat", seat)

        seat = seat + 1

    print()

    row = row + 1
```

## Bank Token Mnaagment system

```python
print("BANK TOKEN MANAGEMENT SYSTEM")
print("------------------------------------------")

bank_name = "Meezan Bank"
branch = "Lahore Main Branch"

total_customers = int(input("Enter Total Customers: "))

print()

token_number = 1

for i in range(1, total_customers + 1):

    print("Customer", i)

    customer_name = input("Enter Customer Name : ")

    print("Token Number        :", token_number)
    print("Branch              :", branch)

    while True:

        print()
        print("Select Service")
        print("1. Cash Deposit")
        print("2. Cash Withdrawal")
        print("3. Account Opening")
        print("4. Balance Inquiry")

        choice = int(input("Enter Your Choice : "))

        if choice == 1:
            service = "Cash Deposit"
            break

        elif choice == 2:
            service = "Cash Withdrawal"
            break

        elif choice == 3:
            service = "Account Opening"
            break

        elif choice == 4:
            service = "Balance Inquiry"
            break

        else:
            print("Invalid Choice! Please Try Again.")

    print()
    print("TOKEN RECEIPT")
    print("------------------------------------------")
    print("Bank Name      :", bank_name)
    print("Customer Name  :", customer_name)
    print("Token Number   :", token_number)
    print("Service        :", service)
    print("Status         : Waiting")
    print("------------------------------------------")

    token_number = token_number + 1

print()
print("All Customers Registered Successfully.")
print("Total Tokens Generated :", total_customers)
print("Thank You For Using", bank_name)
```
