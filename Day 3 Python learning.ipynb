# GROCERY BILL GENERATOR

```python
print("==================================================================")
print("                     GROCERY BILL GENERATOR")
print("==================================================================")

shop_name = "Al Madina Grocery Store"
cashier_name = "Adiba Zia"
invoice_no = "1001"
date = input("Enter Date (DD-MM-YYYY)     : ")

customer_name = input("Enter Customer Name         : ")
phone_number = input("Enter Customer Phone Number : ")

print("\nEnter Item Prices (in Rupees)\n")

rice = int(input("Rice (Rs)                   : "))
flour = int(input("Flour (Rs)                  : "))
sugar = int(input("Sugar (Rs)                  : "))
milk = int(input("Milk (Rs)                   : "))
tea = int(input("Tea (Rs)                    : "))
cooking_oil = int(input("Cooking Oil (Rs)            : "))
eggs = int(input("Eggs (Rs)                   : "))
bread = int(input("Bread (Rs)                  : "))
biscuits = int(input("Biscuits (Rs)               : "))
juice = int(input("Juice (Rs)                  : "))

total_bill = rice + flour + sugar + milk + tea + cooking_oil + eggs + bread + biscuits + juice

print("\n==================================================================")
print("                     GROCERY SHOP BILL")
print("==================================================================")

print("Shop Name          :", shop_name)
print("Invoice Number     :", invoice_no)
print("Date               :", date)
print("Cashier            :", cashier_name)

print("------------------------------------------------------------------")

print("Customer Name      :", customer_name)
print("Phone Number       :", phone_number)

print("------------------------------------------------------------------")

print("Rice               : Rs.", rice)
print("Flour              : Rs.", flour)
print("Sugar              : Rs.", sugar)
print("Milk               : Rs.", milk)
print("Tea                : Rs.", tea)
print("Cooking Oil        : Rs.", cooking_oil)
print("Eggs               : Rs.", eggs)
print("Bread              : Rs.", bread)
print("Biscuits           : Rs.", biscuits)
print("Juice              : Rs.", juice)

print("------------------------------------------------------------------")

print("Total Bill         : Rs.", total_bill)

print("==================================================================")
print("            Thank You For Shopping With Us!")
print("              Visit Again. Have a Nice Day!")
print("==================================================================")
```

```python
print("======================================================================")
print("                    AL MADINA GROCERY STORE"                           )
print("======================================================================")

shop_name = "Al Madina Grocery Store"
cashier = "Adiba Zia"
invoice = "INV-1001"

date = input("Enter Date (DD-MM-YYYY)      : ")
customer = input("Enter Customer Name          : ")
phone = input("Enter Customer Phone Number  : ")

print("\nEnter Item Prices (Rs.)")

rice = int(input("Rice                         : "))
flour = int(input("Flour                        : "))
sugar = int(input("Sugar                        : "))
milk = int(input("Milk                         : "))
tea = int(input("Tea                          : "))
oil = int(input("Cooking Oil                  : "))
eggs = int(input("Eggs                         : "))
bread = int(input("Bread                        : "))
biscuits = int(input("Biscuits                     : "))
juice = int(input("Juice                        : "))

subtotal = rice + flour + sugar + milk + tea + oil + eggs + bread + biscuits + juice

discount = int(input("\nEnter Discount (%)           : "))
tax = int(input("Enter Sales Tax (%)          : "))

discount_amount = subtotal * discount / 100
amount_after_discount = subtotal - discount_amount

tax_amount = amount_after_discount * tax / 100

final_bill = amount_after_discount + tax_amount

print("\n======================================================================")
print("                         PURCHASE INVOICE")
print("======================================================================")

print("Shop Name           :", shop_name)
print("Invoice Number      :", invoice)
print("Date                :", date)
print("Cashier             :", cashier)

print("----------------------------------------------------------------------")

print("Customer Name       :", customer)
print("Phone Number        :", phone)

print("----------------------------------------------------------------------")

print("Rice                : Rs.", rice)
print("Flour               : Rs.", flour)
print("Sugar               : Rs.", sugar)
print("Milk                : Rs.", milk)
print("Tea                 : Rs.", tea)
print("Cooking Oil         : Rs.", oil)
print("Eggs                : Rs.", eggs)
print("Bread               : Rs.", bread)
print("Biscuits            : Rs.", biscuits)
print("Juice               : Rs.", juice)

print("----------------------------------------------------------------------")

print("Subtotal            : Rs.", subtotal)
print("Discount            :", discount, "%")
print("Discount Amount     : Rs.", discount_amount)
print("Amount After Discount : Rs.", amount_after_discount)
print("Sales Tax           :", tax, "%")
print("Tax Amount          : Rs.", tax_amount)

print("----------------------------------------------------------------------")

print("FINAL BILL          : Rs.", final_bill)

print("======================================================================")
print("               Thank You For Shopping With Us")
print("                   Visit Again Soon")
print("======================================================================")
```
